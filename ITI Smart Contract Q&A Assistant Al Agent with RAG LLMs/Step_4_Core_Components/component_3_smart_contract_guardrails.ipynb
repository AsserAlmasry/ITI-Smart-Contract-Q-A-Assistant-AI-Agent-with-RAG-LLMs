```python
# ========================================================
#  COMPONENT 3 - Guardrails (Safety Layer)
# ========================================================

class SmartContractGuardrails:
    """
    Rule-based guardrails for input validation and output moderation.

    Input checks  -> empty/short queries, length limit, prompt injection
    Output checks -> minimum length enforcement, uncertainty detection
    """

    BLOCKED_PATTERNS = [
        r"ignore (previous|all|your) (instructions?|rules?|prompt)",
        r"act as (a )?different (ai|model|assistant)",
        r"jailbreak",
        r"pretend (you are|to be)",
        r"forget (everything|your training)",
        r"(write|generate) (malware|exploit|virus|hack)",
    ]

    def validate_input(self, query: str) -> Tuple[bool, str]:
        """Validate query before processing. Returns (is_valid, reason)."""
        if not query or not query.strip():
            return False, "Query cannot be empty."
        if len(query.strip()) < 5:
            return False, "Query is too short. Please ask a complete question."
        if len(query) > 2000:
            return False, "Query exceeds 2000 character limit."
        for pattern in self.BLOCKED_PATTERNS:
            if re.search(pattern, query.lower()):
                return False, "Query contains a disallowed pattern (prompt injection detected)."
        return True, "OK"

    def validate_output(self, response: str, query: str) -> Tuple[str, List[str]]:
        """Moderate LLM output. Returns (response, warnings_list)."""
        warnings_list = []
        if not response or len(response.strip()) < 20:
            return (
                "Unable to generate a meaningful answer. Please rephrase.",
                ["Empty or very short response detected."]
            )
        uncertainty_phrases = [
            "i'm not sure", "i cannot find", "i don't know", "it's unclear"
        ]
        if any(p in response.lower() for p in uncertainty_phrases):
            warnings_list.append(
                "\u2139\ufe0f Model expressed uncertainty - verify against the source document."
            )
        return response, warnings_list


print("\u2705 SmartContractGuardrails defined.")
```
