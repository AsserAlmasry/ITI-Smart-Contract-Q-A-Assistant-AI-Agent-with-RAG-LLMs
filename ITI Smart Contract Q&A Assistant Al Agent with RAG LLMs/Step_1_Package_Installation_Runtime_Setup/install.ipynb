# 🔐 Smart Contract Summary & Q&A Assistant

**A Production-Ready RAG System for Smart Contract Analysis**

---

This notebook implements a full Retrieval-Augmented Generation (RAG) pipeline:
- 📄 **Ingest** smart contract documents (PDF / DOCX / TXT)
- 🧠 **Embed** and store in ChromaDB via Google Embeddings
- 🔍 **Retrieve** relevant chunks via semantic similarity search
- 💬 **Generate** accurate answers using Gemini 2.5 Flash Lite
- 🛡️ **Guard** against hallucinations and prompt injection
- 🎨 **Interact** through a polished Gradio chat interface
